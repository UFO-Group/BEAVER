```python
import openai
import os
import re
from IPython.display import display, Markdown

# 1. 初始化客户端 (请替换为你自己的真实凭证)
client = openai.OpenAI(
    api_key=os.environ["LLM_API_KEY"], 
    base_url="https://llmapi.paratera.com/v1"
)

# 目标保存路径 (使用 r 前缀处理 Windows 路径)
save_directory = r"..\Judge_Input\Therm._Deg"
file_name = "Qwen3.5-Plus-3.txt"
full_path = os.path.join(save_directory, file_name)

# 2. 定义提示词
user_prompt = (
    "Design a biodegradable polyester with improved heat resistance but without excessively suppressing degradation kinetics. "
)

# 3. 请求模型
print("正在生成设计方案...")
response = client.chat.completions.create(
    model="Qwen3.5-Plus",  # 替换为你要调用的具体模型 ID
    messages=[{"role": "user", "content": user_prompt}]
)

# 获取原始内容
raw_content = response.choices[0].message.content

# 使用正则表达式去除 <think>...</think> 标签及其内部的思考内容
clean_content = re.sub(r'<think>.*?</think>\n*', '', raw_content, flags=re.DOTALL).strip()

# 4. 保存结果到本地文件
try:
    # 如果目录不存在，则创建目录
    if not os.path.exists(save_directory):
        os.makedirs(save_directory)
        print(f"创建目录: {save_directory}")

    # 写入文件，使用清洗后的 clean_content
    with open(full_path, "w", encoding="utf-8") as f:
        f.write(clean_content)
    
    print(f"成功保存纯净版结果至: {full_path}")
except Exception as e:
    print(f"保存文件时出错: {e}")

# 5. 在 Notebook 中预览
display(Markdown("### 模型生成的设计方案（已去除思考过程）："))
display(Markdown(clean_content))
```
