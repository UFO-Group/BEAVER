```python
import os
from pathlib import Path

DATA_ROOT = Path("Data")
from read.rsc import RSCParser  # ✅ 确保有 read/rsc.py 且内有 RSCParser

def parse_rsc_article(filepath, output_folder):
    """解析单篇 RSC HTML 文件并将 paragraph 写入 txt"""
    try:
        # 初始化解析器
        rsc_parser = RSCParser(filepath)
        
        # 解析元信息
        meta_data = rsc_parser.parse_meta()
        print(f"📄 Title: {meta_data.get('title', 'N/A')}")
        print(f"📘 Journal: {meta_data.get('journal', 'N/A')}")
        # print(f"🧾 Abstract: {meta_data.get('abstract', 'N/A')}")
        # print(f"📅 Date: {meta_data.get('date', 'N/A')}")

        # 解析段落（RSCParser.parse_paragraphs 返回的是元素列表）
        para_elements = rsc_parser.parse_paragraphs()
        paragraph_texts = []

        for el in para_elements:
            # lxml 元素：有 text_content 方法
            if hasattr(el, "text_content"):
                txt = el.text_content().strip()
            else:
                txt = str(el).strip()

            if txt:
                paragraph_texts.append(txt)

        print(f"📝 段落数: {len(paragraph_texts)}")

        # 构建输出文件名
        base_name = os.path.basename(filepath).replace(".html", "")
        safe_name = "".join(c for c in base_name if c.isalnum() or c in (" ", "_", "-"))
        output_path = os.path.join(output_folder, f"{safe_name}.txt")

        # 写入到 txt 文件
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(f"Title: {meta_data.get('title', '')}\n")
            f.write(f"Journal: {meta_data.get('journal', '')}\n")
            f.write(f"Date: {meta_data.get('date', '')}\n")
            f.write(f"Abstract: {meta_data.get('abstract', '')}\n\n")
            f.write("Paragraphs:\n")
            for para in paragraph_texts:
                f.write(para + "\n\n")

        print(f"✅ 已保存段落 → {output_path}\n")

    except Exception as e:
        print(f"❌ 解析失败：{filepath}\n错误信息：{e}\n")

def batch_parse_rsc_folder(input_folder, output_folder):
    """批量解析，带【自动跳过】功能"""
    os.makedirs(output_folder, exist_ok=True)
    html_files = [f for f in os.listdir(input_folder) if f.lower().endswith(".html")]

    if not html_files:
        print("⚠️ 未找到 HTML 文件。")
        return

    print(f"🚀 准备处理 {len(html_files)} 篇文献...")

    for html_file in html_files:
        # 1. 预计算输出文件名（必须与 parse_rsc_article 内部逻辑严格一致）
        base_name = html_file.replace(".html", "")
        safe_name = "".join(c for c in base_name if c.isalnum() or c in (" ", "_", "-"))
        output_path = os.path.join(output_folder, f"{safe_name}.txt")

        # 2. 关键判断：如果文件存在且不为空（大于100字节），则跳过
        if os.path.exists(output_path) and os.path.getsize(output_path) > 100:
            # print(f"⏭️  已跳过: {safe_name}") # 如果文件太多可以注释掉这行，避免刷屏
            continue

        # 3. 执行解析
        file_path = os.path.join(input_folder, html_file)
        parse_rsc_article(file_path, output_folder)

    print("🎯 处理任务结束！")

if __name__ == "__main__":
    # 👉 换成你的 RSC HTML 文件夹路径
    input_folder = DATA_ROOT / "RSC" / "source"        # 📂 输入 HTML 文件夹路径
    output_folder = DATA_ROOT / "RSC" / "txt"   # 📂 输出 TXT 文件夹路径
    batch_parse_rsc_folder(input_folder, output_folder)
```
