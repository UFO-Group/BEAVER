```python
import os
from pathlib import Path

DATA_ROOT = Path("Data")
from read.wiley import WileyParser  # ✅ 确保正确导入WileyParser

def parse_wiley_article(filepath, output_folder):
    """解析单篇Wiley HTML文件并将paragraph写入txt"""
    try:
        # 初始化解析器
        wiley_parser = WileyParser(filepath)
        
        # 解析元信息
        meta_data = wiley_parser.parse_meta()
        print(f"📄 Title: {meta_data.get('title', 'N/A')}")
        print(f"📘 Journal: {meta_data.get('journal', 'N/A')}")
        # print(f"🧾 Abstract: {meta_data.get('abstract', 'N/A')}")
        # print(f"📅 Date: {meta_data.get('date', 'N/A')}")
        
        # 解析段落
        paragraphs = wiley_parser.parse_paragraphs()
        print(f"📝 段落数: {len(paragraphs)}")

        # 构建输出文件名
        base_name = os.path.basename(filepath).replace(".html", "")
        safe_name = "".join(c for c in base_name if c.isalnum() or c in (" ", "_", "-"))
        output_path = os.path.join(output_folder, f"{safe_name}.txt")

        # 写入到txt文件
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(f"Title: {meta_data.get('title', '')}\n")
            f.write(f"Journal: {meta_data.get('journal', '')}\n")
            f.write(f"Date: {meta_data.get('date', '')}\n")
            f.write(f"Abstract: {meta_data.get('abstract', '')}\n\n")
            f.write("Paragraphs:\n")
            for para in paragraphs:
                f.write(para.strip() + "\n\n")

        print(f"✅ 已保存段落 → {output_path}\n")

    except Exception as e:
        print(f"❌ 解析失败：{filepath}\n错误信息：{e}\n")


def batch_parse_wiley_folder(input_folder, output_folder):
    """批量解析整个文件夹下的Wiley HTML文件，增加重复跳过逻辑"""
    os.makedirs(output_folder, exist_ok=True)
    html_files = [f for f in os.listdir(input_folder) if f.lower().endswith(".html")]

    if not html_files:
        print("⚠️ 未找到HTML文件，请检查路径。")
        return

    print(f"🚀 开始批量解析，共 {len(html_files)} 篇文献...\n")

    for html_file in html_files:
        # --- 新增：预先构建输出路径来判断是否存在 ---
        base_name = html_file.replace(".html", "")
        safe_name = "".join(c for c in base_name if c.isalnum() or c in (" ", "_", "-"))
        output_path = os.path.join(output_folder, f"{safe_name}.txt")

        if os.path.exists(output_path):
            print(f"⏭️  跳过已处理: {html_file}")
            continue
        # ----------------------------------------

        file_path = os.path.join(input_folder, html_file)
        parse_wiley_article(file_path, output_folder)

    print("🎯 全部解析完成！")

if __name__ == "__main__":
    input_folder = DATA_ROOT / "Wiley" / "source"      # 📂 输入HTML文件夹路径
    output_folder = DATA_ROOT / "Wiley" / "txt"  # 📂 输出TXT文件夹路径
    batch_parse_wiley_folder(input_folder, output_folder)
```
