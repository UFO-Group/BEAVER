```python
from pathlib import Path
from springer import SpringerParser

DATA_ROOT = Path("Data")
# 创建SpringerParser实例
file_path = DATA_ROOT / "examples" / "Springer" / "example.html"
# 创建SpringerParser实例
springer_parser = SpringerParser(file_path)

# 调用parse_meta方法解析文档元数据（标题、摘要、日期、期刊等）
springer_parser.parse_meta()

# 输出文档元数据
print("Title:", springer_parser.title)
print("Journal:", springer_parser.journal)
print("Abstract:", springer_parser.abstract)
print("Date:", springer_parser.date)

# 调用parse_paragraphs方法解析文档中的段落内容
paragraphs = springer_parser.parse_paragraphs()

# 调用parse_tables方法解析表格
springer_parser.parse_tables()  # 解析表格内容，结果可以根据表格内容进一步处理
```
