```python
from pathlib import Path
from wiley import WileyParser  # 确保正确导入WileyParser

DATA_ROOT = Path("Data")

def parse_wiley_article(filepath):
    # 初始化解析器
    wiley_parser = WileyParser(filepath)
    
    # 解析元信息
    meta_data = wiley_parser.parse_meta()
    print(f"Title: {meta_data['title']}")
    print(f"Journal: {meta_data['journal']}")
    print(f"Abstract: {meta_data['abstract']}")
    print(f"Date: {meta_data['date']}")
    
    # 解析段落
    paragraphs = wiley_parser.parse_paragraphs()
    print("Paragraphs:")
    for para in paragraphs:
        print(para)  # 直接输出文本

    # 如果文章包含表格，进行解析
    table_data = wiley_parser.parse_tables()  # 解析所有表格
    print(f"Tables: {table_data}")

# 设置 HTML 文件路径
html_file_path = DATA_ROOT / "examples" / "Wiley" / "example.html"

# 调用解析函数
parse_wiley_article(html_file_path)
```
