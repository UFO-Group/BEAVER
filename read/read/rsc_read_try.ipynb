```python
from pathlib import Path
from rsc import RSCParser   # 假设你将该脚本保存为 rsc.py
from rsc import _rscTableParser   # 假设你将该脚本保存为 rsc.py
from lxml import html
import json

DATA_ROOT = Path("Data")

# 1️⃣ 指定 HTML 文件路径
html_file = DATA_ROOT / "examples" / "RSC" / "example.html"
# 2️⃣ 创建解析器实例
parser = RSCParser(html_file)

# ===========================
# ① 提取元信息
# ===========================
print("\n🧩 --- META INFO ---")
parser.parse_meta()
print(f"📘 Title: {parser.title}")
print(f"📗 Journal: {parser.journal}")
print(f"📅 Date: {parser.date}")
print(f"📄 Abstract: {(parser.abstract or '')}")

paragraphs = parser.parse_paragraphs()

with open(html_file, "rb") as f:
    tree = html.fromstring(f.read())

# 2️⃣ 获取所有表格节点
all_tables = tree.xpath('//table')

print(f"Found {len(all_tables)} tables.")

# 3️⃣ 调用你的解析器
rsc_parser = _rscTableParser()
tables_data = []

for i, t in enumerate(all_tables, start=1):
    parsed = rsc_parser.parse(t)
    if parsed:
        parsed["index"] = i
        tables_data.append(parsed)

# 4️⃣ 打印最终结构
print(json.dumps(tables_data, indent=2, ensure_ascii=False))
```
