```python
from pathlib import Path
from aip import AIPParser

DATA_ROOT = Path("Data")

# ---- 指定文件路径 ----
html_file = DATA_ROOT / "examples" / "AIP" / "example.html"

# ---- 初始化解析器 ----
parser = AIPParser(html_file)

print("\n🧩 --- META INFO ---")
parser.parse_meta()
print(f"📘 Title: {parser.title}")
print(f"📗 Journal: {parser.journal}")
print(f"📅 Date: {parser.date}")
print(f"📄 Abstract: {parser.abstract[:300]}...")  # 截断打印更简洁
print(f"📚 Body text (first 200 chars): {parser.body[:200]}...")


print("\n🧾 --- PARAGRAPHS ---")
paras = parser.parse_paragraphs()


tables = parser.parse_tables()
if not tables:
    print("⚠️ No tables found.")
else:
    print(f"✅ Parsed {len(tables)} tables.")
```
