```python
from pathlib import Path
from elsevier import ElsevierParser

DATA_ROOT = Path("Data")
# Provide the file path of the Elsevier HTML document
file_path = DATA_ROOT / "examples" / "Elsevier" / "example.html"

# Create an instance of the ElsevierParser
elsevier_parser = ElsevierParser(file_path)

# Call the method to parse metadata (title, journal, date, etc.)
elsevier_parser.parse_meta()

# Output the parsed metadata
print(f"Title: {elsevier_parser.title}")
print(f"Journal: {elsevier_parser.journal}")
print(f"Date: {elsevier_parser.date}")
print(f"Body: {elsevier_parser.body}")

# Call the method to parse paragraphs
paragraphs = elsevier_parser.parse_paragraphs()

# Now, call parse_tables() to parse and display tables
elsevier_parser.parse_tables()
```
