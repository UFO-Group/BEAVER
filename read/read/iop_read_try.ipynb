```python
from pathlib import Path
from iop import IOPParser   # 假设脚本文件名是 iop.py

DATA_ROOT = Path("Data")
# 1️⃣ 指定 HTML 文件路径
html_file = DATA_ROOT / "examples" / "IOP" / "example.html"
# 2️⃣ 创建解析器实例
parser = IOPParser(html_file)

# ===========================
# ① 提取元信息
# ===========================
print("\n🧩 --- META INFO ---")
parser.parse_meta()
print(f"📘 Title: {parser.title}")
print(f"📘 Abstract: {parser.abstract}")
print(f"📗 Journal: {parser.journal}")
print(f"📅 Date: {parser.date}")
print(f"📚 Body (first 200 chars): {(parser.body or '')[:200]}...")

paras = parser.parse_paragraphs()

print("\n📊 --- TABLES ---")
parser.parse_tables()
```
